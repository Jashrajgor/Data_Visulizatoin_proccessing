# Handleing missing values with pandas inbuilt methods

```python
df_1=pd.read_csv("raw_data.csv")
```

```python
df_1
```

```python
clean_dataFrame=df_1.copy()
```

```python
clean_dataFrame['name']=clean_dataFrame['name'].fillna("Unknown")
clean_dataFrame['age']=clean_dataFrame['age'].fillna(clean_dataFrame['age'].mean())
clean_dataFrame['country']=clean_dataFrame['country'].ffill()
clean_dataFrame['gender']=clean_dataFrame['gender'].bfill().ffill()
clean_dataFrame['income']=clean_dataFrame['income'].fillna(clean_dataFrame['income'].mean())
```

```python
clean_dataFrame
```

# Handling duplicates

```python
# For duplicate pandas had .duplicated() method that will return a series after search duplicates 
# row wise 
# And to deleting duplicates it had .drop_duplicates() method that will remove whole row from dataframe..
```

# Appying Feature Engineering 

```python
clean_dataFrame['Income_Tax']=clean_dataFrame['income'].apply(lambda x: "20%" if x>=50000 else "10%")
```

```python
clean_dataFrame
```

```python
# map method 
gender_map={"Male":"M","Female":"F","Unknown":"U"}
clean_dataFrame['gender']=clean_dataFrame['gender'].map(gender_map)
```

```python
clean_dataFrame
```

```python
clean_dataFrame.assign(Income_Rating= clean_dataFrame['income'].apply(lambda x:"Good" if x>50000 else "average"))
```

```python
clean_dataFrame.groupby("gender")[['income','Income_Tax']].agg(["max","min"])
```
