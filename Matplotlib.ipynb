# MatplotLib Library for vizulization data on charts

```python
import matplotlib.pyplot as plt  # Here Matplotlib is an library where pyplot is a
# specific module under this library that will help to create charts..
```

```python
x=[1,4,7,10]
y=[5,9,3,9]
```

```python
# Now we will plot this values into the char with the help plot method...
plt.plot(x,y,label="data") # the first argument plot data on x axis and second arguments data ploted on 
# y axis..
plt.title("Example")
# we lables to the x-axis and y-axis with the xlabel() and ylabel() method 
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.legend(loc="upper right") # legend() method only works when label are provided during plot method..
```
