# WEB SCRAPING

```python
# to use we scraping we use two libraries one is 'requests' and onother one is 
# 'BS4 of beautifulSoup' where we can perform scrapin activities on web page....
import requests as req
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# This is the URl from where we fetch the data....
URL="https://www.scrapethissite.com/pages/ajax-javascript/#2010"
```

```python
res=req.get(URL)
```

```python
res.text # This propertry will return the HTML unicode string...
```

```python
# lets store this data into a HTML file so we can furthure use it through that file....
# instead of calling URL number of times...
with open("Scraping.html",'w') as f:
    f.write(res.text)
```

```python
with open("Scraping.html","r") as f:
    HTML_read=f.read()
```

```python
soup=BeautifulSoup(HTML_read,'lxml')
```
