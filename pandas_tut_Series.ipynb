# Pandas Series

```python
# to use pandas we have to install it first, then import it to our programme 
import pandas as pd
```

```python
s=pd.Series([1,2,3,4])
print(f'Pandas series :\n{s}')
```

```python
labeled_series=pd.Series(["370001","380002","350004","390012"],
                         index=["Bhuj","Ghandhidham","Anjar","Mandvi"])
# Like above we had created an series with label, suppose if didn't provide label then by defualt labels are
# corresponding to the index values, so to accessing pandas series we have two options like one is index based and 
# another one label based accessing 
```

```python
# Accessing through the label name are 
print("Pin-code of Bhuj is :",labeled_series['Bhuj']) # like this label i can access the value from series

# Another traditional way is index based that we know "labeled_series[0] it will return the 1st presented value 
# from series...|
```
