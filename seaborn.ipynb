# Seaborn

```python
import seaborn as sns
import pandas as pd
```

```python
penguins=sns.load_dataset("penguins")
penguins
```

```python
data1=pd.DataFrame(penguins["flipper_length_mm"])
data2=pd.DataFrame(penguins['body_mass_g'])
result=pd.concat([data1,data2],axis=1,ignore_index=True)
result.columns=['Flipper_length_mm','Body_mass_g']
# result=penguins[["flipper_length_mm","body_mass_g"]]
result
```

```python
sns.scatterplot(result)
```
