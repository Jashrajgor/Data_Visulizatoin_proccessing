```python
import json
```

```python
def load_data(filename):
    with open(filename,'r') as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
```

```python
def cleaning_data(data):
    rating_num={'one':'1','two':'2','three':'3','four':'4','five':'5'}
    unique_username=set()
    unique_users=[]
    
    for user in data:
        # Correcting all alphabet values with numeric values
        if(user['rating'] in rating_num):
            user['rating']=rating_num[user['rating']]
            
        # Adding none vlaue to the missing values
        user['age']=user.get('age')

        #Removing the duplicate values
        if(user['name'] in unique_username):
            continue
        unique_username.add(user['name'])
        unique_users.append(user)
    return unique_users
        
```

```python
json_data=cleaning_data(data)
```

```python
def json_dataWriting(pyData):
    with open("new_json.json",'w') as f:
        f.write(json.dumps(json_data))
```

```python
json_dataWriting(json_data)
```

```python
# Extracting the meaning full insights from the data
def get_insights(data):
    
    #Declaring function variables
    avg_rating=0
    poor_rating=0
    
    # Calculatin average of rating
    for i in data:
        avg_rating+=float((i['rating']))
        
    avg_rating=avg_rating/len(data)
    print(f'Average rating is : {avg_rating}')

    # Calculating user's with poor rating less than 3 
    for j in data:
        if(float(j['rating'])<3):
            poor_rating+=1
    print(f"Percentage of poor rating is : {poor_rating/len(data)*100}")
    
```

```python
get_insights(json_data)
```
