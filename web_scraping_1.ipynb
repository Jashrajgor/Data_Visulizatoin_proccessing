```python
import requests as req
from bs4 import BeautifulSoup
import pandas as pd
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.9',
}
count=1
title_author=[]
while count<5:
        res=req.get(f'http://quotes.toscrape.com/page/{count}/',headers=headers)
        if(res.status_code==200):
            with open(f'Qutes.html_{count}','w',encoding='utf-8') as f:
                f.write(res.text)
                count=count+1
```

```python
for i in range(1,count):
    with open(f'Qutes.html_{i}','r',encoding='utf-8') as f:
        html_content=f.read()
        soup=BeautifulSoup(html_content,'lxml')
        for tags in soup.select("a.tag"):
            if(tags.text=="life"):
                parent_tags=tags.find_parent(class_="quote")
                if(parent_tags!=None):
                    title=parent_tags.find("span").text
                    author=parent_tags.select_one(".author").text
                    title_author.append([title,author])
```

```python
dataFrame=pd.DataFrame(title_author,columns=['Title',"Author"])
```
